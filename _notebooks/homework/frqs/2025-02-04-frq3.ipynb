# 2024 FRQ 2 - Scoreboard

## Problem Overview

The 2024 AP CSA FRQ #2 asks us to implement a `Scoreboard` class that tracks the score and active team for a two-team game. The class needs to:

1. **Store information** about two teams, their scores, and which team is currently active
2. **Record plays** that either score points (keeping the same team active) or fail (switching to the other team)
3. **Return the current game state** as a formatted string

## My Implementation

`// CODE_RUNNER: Answer`


````java
// CODE_RUNNER: Answer  

public class Main {
    public static void main(String[] args) {
        String info;
        
        // Create a new scoreboard
        Scoreboard game = new Scoreboard("Red", "Blue");
        
        // Test sequence from FRQ
        info = game.getScore();
        System.out.println("Initial: " + info);  // 0-0-Red
        
        game.recordPlay(1);
        info = game.getScore();
        System.out.println("After Red scores 1: " + info);  // 1-0-Red
        
        game.recordPlay(0);
        info = game.getScore();
        System.out.println("After Red fails: " + info);  // 1-0-Blue
        
        game.recordPlay(3);
        info = game.getScore();
        System.out.println("After Blue scores 3: " + info);  // 1-3-Blue
        
        game.recordPlay(1);
        game.recordPlay(0);
        info = game.getScore();
        System.out.println("After Blue scores 1 then fails: " + info);  // 1-4-Red
        
        game.recordPlay(0);
        game.recordPlay(4);
        game.recordPlay(0);
        info = game.getScore();
        System.out.println("Final: " + info);  // 1-8-Red
        
        // Test with different teams
        Scoreboard match = new Scoreboard("Lions", "Tigers");
        info = match.getScore();
        System.out.println("New match: " + info);  // 0-0-Lions
        
        // Original game state unchanged
        info = game.getScore();
        System.out.println("Original game still: " + info);  // 1-8-Red
    }
}

class Scoreboard {
    // Instance variables to track game state
    private String team1;
    private String team2;
    private int score1;
    private int score2;
    private boolean team1Active;
    
    // Constructor initializes the game with two team names
    // Team 1 always starts as the active team
    public Scoreboard(String t1, String t2) {
        team1 = t1;
        team2 = t2;
        score1 = 0;
        score2 = 0;
        team1Active = true;  // Team 1 starts active
    }
    
    // Records a play: positive points = scoring play, 0 points = failed play
    public void recordPlay(int points) {
        if (points > 0) {
            // Scoring play: add points to active team's score
            if (team1Active) {
                score1 += points;
            } else {
                score2 += points;
            }
            // Active team remains the same
        } else {
            // Failed play (0 points): switch active team
            team1Active = !team1Active;
        }
    }
    
    // Returns current score in format "score1-score2-activeTeam"
    public String getScore() {
        String activeTeam;
        if (team1Active) {
            activeTeam = team1;
        } else {
            activeTeam = team2;
        }
        return score1 + "-" + score2 + "-" + activeTeam;
    }
}

Main.main(null);
````

## Code Explanation

### Instance Variables
- `team1`, `team2`: Store the names of the two teams
- `score1`, `score2`: Track each team's current score
- `team1Active`: Boolean flag indicating which team is currently active (true = team1, false = team2)

### Constructor
The constructor takes two team names and initializes:
- Team names to the provided parameters
- Both scores to 0
- `team1Active` to true (team 1 always starts)

### recordPlay(int points) Method
This is the core logic:
- **If points > 0**: It's a scoring play
  - Add points to the active team's score
  - Active team stays the same (they get another turn)
- **If points == 0**: It's a failed play
  - Switch the active team using boolean negation (`!team1Active`)
  - No points are added

### getScore() Method
Returns a formatted string showing:
1. Team 1's score
2. Team 2's score  
3. The currently active team's name

The method determines which team is active and returns the information in "score1-score2-activeTeam" format.

## My Struggle

The main challenge I faced was **understanding when the active team should change**. At first, I wasn't sure if:
- The team should alternate after every play (like turns in a board game)
- The team should only change on failed plays

By carefully reading the problem and working through the examples, I realized:
- **Successful plays (points > 0)**: The same team continues playing
- **Failed plays (points == 0)**: Control switches to the other team

This reflects real sports where a team keeps possession as long as they're scoring, but loses it when they fail to score. Once I understood this concept, implementing the `recordPlay` method became straightforward using a simple if-else statement.
