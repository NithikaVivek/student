## Popcorn Hack 1.2: Type Safety
Question: What happens if you write int count = "hello";? Why does this fail, and why is this behavior actually helpful?
Answer:
This code will cause a compilation error with a message like "incompatible types: String cannot be converted to int."
Why it fails:

- int can only store integer values (whole numbers)
- "hello" is a String literal (text)
- Java is a strongly-typed language, meaning variable types are strictly enforced

Why this is helpful:

- Catches errors early at compile-time rather than runtime
- Prevents logical bugs (e.g., trying to do math operations on text)
- Makes code more predictable and easier to debug
- Forces programmers to think carefully about data types

## Popcorn Hack 2.1: Input Validation
Question: What happens if a user types "twenty" when nextInt() expects a number? How would you make your program more robust?

Answer: The program will throw an InputMismatchException and crash.

How to make it more robust:


```java
Scanner sc = new Scanner(System.in);
int age = 0;
boolean validInput = false;

while (!validInput) {
    System.out.print("Enter your age: ");
    if (sc.hasNextInt()) {
        age = sc.nextInt();
        validInput = true;
        System.out.print(age);
    } else {
        System.out.println("Invalid input! Please enter a number.");
        sc.next(); // Clear the invalid input
    }
}
```

## Popcorn Hack 2.2: Token vs Line

**Question:** If a user enters "San Diego" when `next()` is called, what gets stored? What about with `nextLine()`?

**Answer:**

**With `next()`:**
- Only `"San"` gets stored
- `next()` reads only up to the first whitespace
- `"Diego"` remains in the buffer for the next read

**With `nextLine()`:**
- `"San Diego"` gets stored (the entire line)
- `nextLine()` reads everything until the newline character (`\n`)

## Homework Hack 1

```java
import java.util.Scanner;

public class AverageCalculator {
    public static void main(String[] args) {
        System.out.println("=== Three-Number Average Calculator ===\n");
        Scanner sc = new Scanner(System.in);
        
        // Prompt for three integers
        System.out.print("Enter the first number: ");
        int num1 = sc.nextInt();
        
        System.out.print("Enter the second number: ");
        int num2 = sc.nextInt();
        
        System.out.print("Enter the third number: ");
        int num3 = sc.nextInt();
        
        // Calculate average (cast to double to avoid integer division)
        double average = (double) (num1 + num2 + num3) / 3;
        
        // Display result with two decimal places
        System.out.println("\n--- Results ---");
        System.out.println("Number 1: " + num1);
        System.out.println("Number 2: " + num2);
        System.out.println("Number 3: " + num3);
        System.out.println("Sum: " + (num1 + num2 + num3));
        System.out.printf("Average: %.2f\n", average);
        
        sc.close();
    }
}

AverageCalculator.main(null);
```

## Homework Hack 2

Assignment Operator (=)
The assignment operator = is used to store a value into a variable.
Key Points:
- Works right to left (evaluates right side first, then stores in left side)
- Changes the value of the variable
- Used for initialization and updating variables
- Returns the assigned value

Comparison Operator (==)
The comparison operator == is used to test if two values are equal.
Key Points:
- Returns a boolean value (true or false)
- Does NOT change any variables
- Used in conditional statements (if, while, etc.)
- Compares VALUES for primitives, REFERENCES for objects
