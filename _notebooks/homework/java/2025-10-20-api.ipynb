```java
import java.util.ArrayList;

public class PopcornHack1 {
    public static void main(String[] args) {
        // Use Math.pow() to calculate 3^4
        double powerResult = Math.pow(3, 4);
        System.out.println("3^4 = " + powerResult);
        
        // Use Math.sqrt() to find square root of 64
        double sqrtResult = Math.sqrt(64);
        System.out.println("Square root of 64 = " + sqrtResult);
        
        // Create ArrayList of Strings
        ArrayList<String> colors = new ArrayList<>();
        
        // Add 3 colors ("red", "blue", "green")
        colors.add("red");
        colors.add("blue");
        colors.add("green");
        
        // Print the ArrayList and its size
        System.out.println("Colors: " + colors);
        System.out.println("Number of colors: " + colors.size());
    }
}

PopcornHack1.main(null);
```

```java
public class Book {
    // Attributes - what the book HAS
    private String title;
    private String author;
    private int pages;
    
    // Constructor to set all attributes
    public Book(String title, String author, int pages) {
        this.title = title;
        this.author = author;
        this.pages = pages;
    }
    
    // Behavior: displayInfo() - prints all book information
    public void displayInfo() {
        System.out.println("Title: " + title);
        System.out.println("Author: " + author);
        System.out.println("Pages: " + pages);
        System.out.println("---");
    }
    
    // Behavior: isLong() - returns true if pages > 300
    public boolean isLong() {
        return pages > 300;
    }
    
    // Main method to test the Book class
    public static void main(String[] args) {
        // Create first Book object
        Book myBook = new Book("Java Basics", "John Doe", 350);
        
        // Test all methods
        myBook.displayInfo();
        System.out.println("Is this book long? " + myBook.isLong());
        
        System.out.println(); // blank line for separation
        
        // Create another Book object to test with a short book
        Book shortBook = new Book("Quick Guide to Python", "Jane Smith", 150);
        shortBook.displayInfo();
        System.out.println("Is this book long? " + shortBook.isLong());
    }
}

Book.main(null);
```

```java
import java.util.ArrayList;

public class Phone {
    // Attributes (4) - what the phone HAS
    private String brand;
    private String model;
    private int batteryLevel;
    private ArrayList<String> contacts;
    
    // Constructor - sets brand and model, initializes empty contacts list
    public Phone(String brand, String model) {
        this.brand = brand;
        this.model = model;
        this.batteryLevel = 100; // starts at 100%
        this.contacts = new ArrayList<>();
    }
    
    // Behavior 1: displayInfo() - prints brand, model, and battery level
    public void displayInfo() {
        System.out.println("=== Phone Information ===");
        System.out.println("Brand: " + brand);
        System.out.println("Model: " + model);
        System.out.println("Battery Level: " + batteryLevel + "%");
        System.out.println("Number of Contacts: " + contacts.size());
        System.out.println("========================");
    }
    
    // Behavior 2: addContact(String name) - adds name to contacts list
    public void addContact(String name) {
        contacts.add(name);
        System.out.println("Added contact: " + name);
    }
    
    // Behavior 3: showContacts() - prints all contacts
    public void showContacts() {
        System.out.println("\n--- Contacts List ---");
        if (contacts.isEmpty()) {
            System.out.println("No contacts saved.");
        } else {
            for (int i = 0; i < contacts.size(); i++) {
                System.out.println((i + 1) + ". " + contacts.get(i));
            }
        }
        System.out.println("--------------------");
    }
    
    // Behavior 4: usePhone(int minutes) - decreases battery by minutes used
    public void usePhone(int minutes) {
        batteryLevel -= minutes;
        if (batteryLevel < 0) {
            batteryLevel = 0;
        }
        System.out.println("Used phone for " + minutes + " minutes. Battery now at " + batteryLevel + "%");
    }
}

// Test the Phone class
class PhoneTest {
    public static void main(String[] args) {
        // Create 2 Phone objects
        Phone phone1 = new Phone("Apple", "iPhone 15");
        Phone phone2 = new Phone("Samsung", "Galaxy S24");
        
        System.out.println("========== PHONE 1 ==========\n");
        
        // Add 3 contacts to phone1
        phone1.addContact("Alice Johnson");
        phone1.addContact("Bob Smith");
        phone1.addContact("Charlie Brown");
        
        // Show contacts for phone1
        phone1.showContacts();
        
        // Use phone1 for some minutes
        phone1.usePhone(25);
        phone1.usePhone(15);
        
        // Display phone1 information
        System.out.println();
        phone1.displayInfo();
        
        System.out.println("\n========== PHONE 2 ==========\n");
        
        // Add 3 contacts to phone2
        phone2.addContact("Diana Prince");
        phone2.addContact("Evan Williams");
        phone2.addContact("Fiona Green");
        
        // Show contacts for phone2
        phone2.showContacts();
        
        // Use phone2 for some minutes
        phone2.usePhone(30);
        phone2.usePhone(20);
        
        // Display phone2 information
        System.out.println();
        phone2.displayInfo();
        
        System.out.println("\n========== FINAL COMPARISON ==========\n");
        System.out.println("Phone 1 Contacts:");
        phone1.showContacts();
        System.out.println("\nPhone 2 Contacts:");
        phone2.showContacts();
    }
}

PhoneTest.main(null);
```
