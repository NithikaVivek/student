## Popcorn Hack: Instance vs Static Methods
Look at the code below and decide which lines will work and which will cause an error. Then fix the wrong ones.

```java
class Bro {
    String name;
    static String catchphrase = "Sup bro?";
    public Bro(String name) {
        this.name = name;
    }
    public void sayHi() {
        System.out.println("Hey, I'm " + name);
    }
    public static void sayCatchphrase() {
        System.out.println(catchphrase);
    }
    public static void main(String[] args) {
        sayCatchphrase(); // Works
        // sayHi(); // Error: non-static method in static context
        Bro alex = new Bro("Alex");
        alex.sayHi(); // Works
        alex.sayCatchphrase(); // Works, but not recommended (should use Bro.sayCatchphrase())
    }
}
```

**Error Explanation:**
- `sayHi();` causes an error because it's an instance method called from a static context.
- To fix, call `sayHi()` on an object, e.g., `alex.sayHi();`.

## Popcorn Hack Solution: Fixed Code
Here is the corrected version of the Bro class for the Popcorn Hack. The instance method is now called on an object, and the static method is called using the class name.

**Summary of Fixes:**
- Call `sayCatchphrase()` using `Bro.sayCatchphrase()`.
- Call `sayHi()` using an object, e.g., `alex.sayHi();`.
- Do not call instance methods from a static context without an object.

````java
class Bro {
    String name;
    static String catchphrase = "Sup bro?";
    public Bro(String name) {
        this.name = name;
    }
    public void sayHi() {
        System.out.println("Hey, I'm " + name);
    }
    public static void sayCatchphrase() {
        System.out.println(catchphrase);
    }
    public static void main(String[] args) {
        Bro.sayCatchphrase(); // Correct: static method called with class name
        // Bro.sayHi(); // Error: cannot call instance method from class
        Bro alex = new Bro("Alex");
        alex.sayHi(); // Correct: instance method called on object
        alex.sayCatchphrase(); // Works, but Bro.sayCatchphrase() is preferred
    }
}

Bro.main(null);
````

````java
// Ultimate Battle: Robot vs Dinosaur
class Fighter {
    String name;
    int power;
    int health;
    static double fightDuration = 0;
    
    public Fighter(String name, int power, int health) {
        this.name = name;
        this.power = power;
        this.health = health;
    }
    
    // Instance method: attack another fighter
    public void attack(Fighter opponent) {
        System.out.println(this.name + " attacks " + opponent.name + "!");
        opponent.health -= this.power;
        fightDuration += 1.5; // Each attack adds to fight duration
    }
    
    // Instance method: print status
    public void printStatus() {
        System.out.println(name + " | Health: " + health + " | Power: " + power);
    }
    
    // Instance method: taunt
    public void taunt() {
        System.out.println(name + " says: Bring it on!");
    }
    
    // Static method: compare power
    public static int strongerFighter(Fighter f1, Fighter f2) {
        if (f1.power > f2.power) return 1;
        else if (f2.power > f1.power) return 2;
        else return 0;
    }
    
    // Static method: begin battle
    public static void beginBattle(Fighter f1, Fighter f2) {
        System.out.println("Battle begins between " + f1.name + " and " + f2.name + "!");
        f1.taunt();
        f2.taunt();
        while (f1.health > 0 && f2.health > 0) {
            f1.attack(f2);
            if (f2.health <= 0) break;
            f2.attack(f1);
        }
        System.out.println("Fight duration: " + fightDuration + " seconds");
        if (f1.health > 0) System.out.println(f1.name + " wins!");
        else System.out.println(f2.name + " wins!");
    }
    
    // Static method: fun fact
    public static void printFact() {
        System.out.println("Fun Fact: The stronger fighter doesn't always win!");
    }
    
    public static void main(String[] args) {
        Fighter robot = new Fighter("Robot", 30, 100);
        Fighter dino = new Fighter("Dinosaur", 25, 120);
        
        robot.printStatus();
        dino.printStatus();
        
        // Compare power
        int result = Fighter.strongerFighter(robot, dino);
        if (result == 1) System.out.println(robot.name + " is stronger!");
        else if (result == 2) System.out.println(dino.name + " is stronger!");
        else System.out.println("Both fighters are equally strong!");
        
        Fighter.printFact();
        
        // Begin battle
        Fighter.beginBattle(robot, dino);
        
        // Print final status
        robot.printStatus();
        dino.printStatus();
    }
}
Fighter.main(null);
````
