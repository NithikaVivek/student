# FRQ Coding Tasks: Java Casting & Range

Below are solutions to the three FRQ-style coding problems from Unit 1.5. Each method is explained and demonstrated in code.

```java
// FRQ 1: Average with correct casting
public static double avgInt(int a, int b) {
    // Convert one operand to double to avoid integer division truncation
    return ((double) a + b) / 2.0;
}

// Print results directly
System.out.println("Average of 5 and 6 = " + avgInt(5, 6));  // Expected: 5.5
System.out.println("Average of 2 and 3 = " + avgInt(2, 3));  // Expected: 2.5
```

```java
// FRQ 2: Percentage calculation
public static double percent(int correct, int total) {
    // Prevent division by zero
    if (total == 0) return 0.0;
    // Compute percentage as a precise double
    return 100.0 * ((double) correct) / total;
}

// Print results directly
System.out.println("Percent: " + percent(47, 50));  // Expected: 94.0
System.out.println("Percent: " + percent(0, 0));    // Expected: 0.0
```

```java
// FRQ 3: Safe remainder
public static int safeMod(int a, int b) {
    // Return 0 instead of throwing an exception if b == 0
    if (b == 0) return 0;
    return a % b;
}

// Print results directly
System.out.println("safeMod(10, 3) = " + safeMod(10, 3));  // Expected: 1
System.out.println("safeMod(10, 0) = " + safeMod(10, 0));  // Expected: 0
```
