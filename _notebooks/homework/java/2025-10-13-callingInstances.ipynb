## POPCORN HACK #1: Enhanced Dog Class

```java
class Dog {
    private String name;
    private String breed;
    
    public Dog(String name, String breed) {
        this.name = name;
        this.breed = breed;
    }
    
    public void bark() {
        System.out.println(name + " says: Woof!");
    }
    
    public void printDescription() {
        System.out.println("My dog " + name + " is a " + breed + "!");
    }
}

Dog d = new Dog("Fido", "Golden Retriever");
d.bark();
d.printDescription();
```

## POPCORN HACK #2: Enhanced Counter Class


```java
class Counter {
    private int count;
    
    public void add(int x) {
        count += x;
    }
    
    public void subtract(int x) {
        count -= x;
    }
    
    public void multiply(int x) {
        count *= x;
    }
    
    public void divide(int x) {
        if (x != 0) {
            count /= x;
        } else {
            System.out.println("Cannot divide by zero!");
        }
    }
    
    public int getCount() {
        return count;
    }
}

// Usage:
Counter c = new Counter();
c.add(10);
System.out.println("After add(10): " + c.getCount());
c.subtract(3);
System.out.println("After subtract(3): " + c.getCount());
c.multiply(2);
System.out.println("After multiply(2): " + c.getCount());
c.divide(2);
System.out.println("After divide(2): " + c.getCount());
```

## HOMEWORK: Student Grade Tracker

```java
public class StudentGradeTracker {
    public static void main(String[] args) {
        System.out.println("=== Student Grade Tracker System ===\n");
        
        // Create first student
        System.out.println("Creating student: Emma Rodriguez");
        Student emma = new Student("Emma Rodriguez");
        System.out.println("Student created successfully!\n");
        
        // Add grades for Emma
        System.out.println("--- Adding Grades for Emma ---");
        emma.addGrade(95);
        System.out.println("Adding grade: 95 points");
        emma.addGrade(88);
        System.out.println("Adding grade: 88 points");
        emma.addGrade(92);
        System.out.println("Adding grade: 92 points");
        emma.addGrade(85);
        System.out.println("Adding grade: 85 points");
        
        // Print Emma's report
        System.out.println("\n--- Emma's Grade Report ---");
        emma.printReport();
        
        System.out.println("\n========================================\n");
        
        // Create second student
        System.out.println("Creating student: James Wilson");
        Student james = new Student("James Wilson");
        System.out.println("Student created successfully!\n");
        
        // Add grades for James
        System.out.println("--- Adding Grades for James ---");
        james.addGrade(78);
        System.out.println("Adding grade: 78 points");
        james.addGrade(82);
        System.out.println("Adding grade: 82 points");
        james.addGrade(75);
        System.out.println("Adding grade: 75 points");
        
        // Print James's report
        System.out.println("\n--- James's Grade Report ---");
        james.printReport();
        
        System.out.println("\n========================================\n");
        
        // Final summary
        System.out.println("Final Summary:");
        System.out.printf("Emma Rodriguez - Average: %.2f (%s)\n", emma.getAverage(), emma.getLetterGrade());
        System.out.printf("James Wilson - Average: %.2f (%s)\n", james.getAverage(), james.getLetterGrade());
    }
}

class Student {
    // Instance variables
    private String name;
    private int totalPoints;
    private int numAssignments;
    
    // Constructor
    public Student(String name) {
        this.name = name;
        this.totalPoints = 0;
        this.numAssignments = 0;
    }
    
    // Void method: adds a grade to the student's record
    public void addGrade(int points) {
        totalPoints += points;
        numAssignments++;
    }
    
    // Non-void method: calculates and returns average
    public double getAverage() {
        if (numAssignments == 0) {
            return 0.0;
        }
        return (double) totalPoints / numAssignments;
    }
    
    // Non-void method: returns letter grade based on average
    public String getLetterGrade() {
        double average = getAverage();
        if (average >= 90) {
            return "A";
        } else if (average >= 80) {
            return "B";
        } else if (average >= 70) {
            return "C";
        } else if (average >= 60) {
            return "D";
        } else {
            return "F";
        }
    }
    
    // Void method: prints a formatted grade report
    public void printReport() {
        System.out.println("Student Name: " + name);
        System.out.println("Total Points: " + totalPoints);
        System.out.println("Assignments Completed: " + numAssignments);
        System.out.printf("Current Average: %.2f\n", getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());
        
        // Add motivational status message
        String status;
        String letterGrade = getLetterGrade();
        if (letterGrade.equals("A")) {
            status = "Excellent work!";
        } else if (letterGrade.equals("B")) {
            status = "Good job!";
        } else if (letterGrade.equals("C")) {
            status = "Keep working hard!";
        } else if (letterGrade.equals("D")) {
            status = "Need improvement!";
        } else {
            status = "Seek help immediately!";
        }
        System.out.println("Status: " + status);
    }
}
```

```java
StudentGradeTracker.main(null);
```

## Extra Credit

```java
import java.util.ArrayList;

public class StudentGradeTrackerExtra {
    public static void main(String[] args) {
        System.out.println("=== Enhanced Student Grade Tracker System ===\n");
        
        // Create first student
        System.out.println("Creating student: Emma Rodriguez");
        Student emma = new Student("Emma Rodriguez");
        System.out.println("Student created successfully!\n");
        
        // Add grades for Emma
        System.out.println("--- Adding Grades for Emma ---");
        emma.addGrade(95);
        System.out.println("Adding grade: 95 points");
        emma.addGrade(88);
        System.out.println("Adding grade: 88 points");
        emma.addGrade(92);
        System.out.println("Adding grade: 92 points");
        emma.addGrade(85);
        System.out.println("Adding grade: 85 points");
        
        // Print Emma's report
        System.out.println("\n--- Emma's Grade Report ---");
        emma.printReport();
        
        System.out.println("\n========================================\n");
        
        // Create second student
        System.out.println("Creating student: James Wilson");
        Student james = new Student("James Wilson");
        System.out.println("Student created successfully!\n");
        
        // Add grades for James
        System.out.println("--- Adding Grades for James ---");
        james.addGrade(78);
        System.out.println("Adding grade: 78 points");
        james.addGrade(82);
        System.out.println("Adding grade: 82 points");
        james.addGrade(75);
        System.out.println("Adding grade: 75 points");
        
        // Print James's report
        System.out.println("\n--- James's Grade Report ---");
        james.printReport();
        
        System.out.println("\n========================================\n");
        
        // Demonstrate bonus features
        System.out.println("=== BONUS FEATURES DEMONSTRATION ===\n");
        
        // Highest and Lowest grades
        System.out.println("--- Highest/Lowest Grades ---");
        System.out.println("Emma's highest grade: " + emma.getHighestGrade());
        System.out.println("Emma's lowest grade: " + emma.getLowestGrade());
        System.out.println("James's highest grade: " + james.getHighestGrade());
        System.out.println("James's lowest grade: " + james.getLowestGrade());
        
        // Compare students
        System.out.println("\n--- Comparing Students ---");
        int comparison = emma.compareTo(james);
        if (comparison > 0) {
            System.out.println("Emma has a higher average than James");
        } else if (comparison < 0) {
            System.out.println("James has a higher average than Emma");
        } else {
            System.out.println("Emma and James have the same average");
        }
        
        // Drop lowest grade
        System.out.println("\n--- Testing Drop Lowest Grade ---");
        System.out.printf("James's average before dropping lowest: %.2f\n", james.getAverage());
        james.dropLowestGrade();
        System.out.println("Dropped James's lowest grade!");
        System.out.printf("James's average after dropping lowest: %.2f\n", james.getAverage());
        james.printReport();
        
        // Weighted grades
        System.out.println("\n--- Testing Weighted Grades ---");
        Student sarah = new Student("Sarah Chen");
        sarah.addGrade(85, 1.0);  // Regular assignment
        sarah.addGrade(90, 1.0);  // Regular assignment
        sarah.addGrade(95, 2.0);  // Weighted 2x (like a test)
        System.out.println("Added grades with weights: 85 (1x), 90 (1x), 95 (2x)");
        sarah.printReport();
        
        // Reset feature
        System.out.println("\n--- Testing Reset Feature ---");
        Student temp = new Student("Temp Student");
        temp.addGrade(100);
        temp.addGrade(90);
        System.out.println("Before reset:");
        temp.printReport();
        temp.reset();
        System.out.println("\nAfter reset:");
        temp.printReport();
        
        // Error checking
        System.out.println("\n--- Testing Error Checking ---");
        System.out.println("Attempting to add negative grade (-50):");
        temp.addGrade(-50);
        System.out.println("Attempting to add grade over 100 (150):");
        temp.addGrade(150);
        
        System.out.println("\n========================================\n");
        
        // Final summary
        System.out.println("Final Summary:");
        System.out.printf("Emma Rodriguez - Average: %.2f (%s)\n", emma.getAverage(), emma.getLetterGrade());
        System.out.printf("James Wilson - Average: %.2f (%s)\n", james.getAverage(), james.getLetterGrade());
        System.out.printf("Sarah Chen - Average: %.2f (%s)\n", sarah.getAverage(), sarah.getLetterGrade());
    }
}

class Student {
    // Instance variables
    private String name;
    private ArrayList<Integer> grades;
    private ArrayList<Double> weights;
    
    // Constructor
    public Student(String name) {
        this.name = name;
        this.grades = new ArrayList<>();
        this.weights = new ArrayList<>();
    }
    
    // Original void method: adds a grade with default weight of 1.0
    public void addGrade(int points) {
        addGrade(points, 1.0);
    }
    
    // Enhanced void method: adds a grade with custom weight (error checking included)
    public void addGrade(int points, double weight) {
        if (points < 0) {
            System.out.println("Error: Grade cannot be negative. Grade not added.");
            return;
        }
        if (points > 100) {
            System.out.println("Error: Grade cannot exceed 100. Grade not added.");
            return;
        }
        if (weight <= 0) {
            System.out.println("Error: Weight must be positive. Grade not added.");
            return;
        }
        grades.add(points);
        weights.add(weight);
    }
    
    // Non-void method: calculates and returns weighted average
    public double getAverage() {
        if (grades.isEmpty()) {
            return 0.0;
        }
        double totalWeightedPoints = 0.0;
        double totalWeight = 0.0;
        for (int i = 0; i < grades.size(); i++) {
            totalWeightedPoints += grades.get(i) * weights.get(i);
            totalWeight += weights.get(i);
        }
        return totalWeightedPoints / totalWeight;
    }
    
    // Non-void method: returns letter grade based on average
    public String getLetterGrade() {
        double average = getAverage();
        if (average >= 90) {
            return "A";
        } else if (average >= 80) {
            return "B";
        } else if (average >= 70) {
            return "C";
        } else if (average >= 60) {
            return "D";
        } else {
            return "F";
        }
    }
    
    // BONUS: Non-void method to get highest grade
    public int getHighestGrade() {
        if (grades.isEmpty()) {
            return 0;
        }
        int highest = grades.get(0);
        for (int grade : grades) {
            if (grade > highest) {
                highest = grade;
            }
        }
        return highest;
    }
    
    // BONUS: Non-void method to get lowest grade
    public int getLowestGrade() {
        if (grades.isEmpty()) {
            return 0;
        }
        int lowest = grades.get(0);
        for (int grade : grades) {
            if (grade < lowest) {
                lowest = grade;
            }
        }
        return lowest;
    }
    
    // BONUS: Void method to drop the lowest grade
    public void dropLowestGrade() {
        if (grades.isEmpty()) {
            System.out.println("No grades to drop.");
            return;
        }
        int lowestIndex = 0;
        int lowestGrade = grades.get(0);
        for (int i = 1; i < grades.size(); i++) {
            if (grades.get(i) < lowestGrade) {
                lowestGrade = grades.get(i);
                lowestIndex = i;
            }
        }
        grades.remove(lowestIndex);
        weights.remove(lowestIndex);
        System.out.println("Dropped grade: " + lowestGrade);
    }
    
    // BONUS: Non-void method to compare students by average
    public int compareTo(Student other) {
        double thisAverage = this.getAverage();
        double otherAverage = other.getAverage();
        if (thisAverage > otherAverage) {
            return 1;
        } else if (thisAverage < otherAverage) {
            return -1;
        } else {
            return 0;
        }
    }
    
    // BONUS: Void method to reset all grades
    public void reset() {
        grades.clear();
        weights.clear();
        System.out.println("All grades cleared for " + name);
    }
    
    // Void method: prints a formatted grade report
    public void printReport() {
        System.out.println("Student Name: " + name);
        System.out.println("Assignments Completed: " + grades.size());
        if (!grades.isEmpty()) {
            System.out.println("Highest Grade: " + getHighestGrade());
            System.out.println("Lowest Grade: " + getLowestGrade());
        }
        System.out.printf("Current Average: %.2f\n", getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());
        
        // Add motivational status message
        String status;
        String letterGrade = getLetterGrade();
        if (letterGrade.equals("A")) {
            status = "Excellent work!";
        } else if (letterGrade.equals("B")) {
            status = "Good job!";
        } else if (letterGrade.equals("C")) {
            status = "Keep working hard!";
        } else if (letterGrade.equals("D")) {
            status = "Need improvement!";
        } else {
            status = "Seek help immediately!";
        }
        System.out.println("Status: " + status);
    }
}

StudentGradeTrackerExtra.main(null);
```
