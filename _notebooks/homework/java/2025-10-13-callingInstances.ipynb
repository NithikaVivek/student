## POPCORN HACK #1: Enhanced Dog Class

```java
class Dog {
    private String name;
    private String breed;
    
    public Dog(String name, String breed) {
        this.name = name;
        this.breed = breed;
    }
    
    public void bark() {
        System.out.println(name + " says: Woof!");
    }
    
    public void printDescription() {
        System.out.println("My dog " + name + " is a " + breed + "!");
    }
}

Dog d = new Dog("Fido", "Golden Retriever");
d.bark();
d.printDescription();
```

## POPCORN HACK #2: Enhanced Counter Class


```java
class Counter {
    private int count;
    
    public void add(int x) {
        count += x;
    }
    
    public void subtract(int x) {
        count -= x;
    }
    
    public void multiply(int x) {
        count *= x;
    }
    
    public void divide(int x) {
        if (x != 0) {
            count /= x;
        } else {
            System.out.println("Cannot divide by zero!");
        }
    }
    
    public int getCount() {
        return count;
    }
}

// Usage:
Counter c = new Counter();
c.add(10);
System.out.println("After add(10): " + c.getCount());
c.subtract(3);
System.out.println("After subtract(3): " + c.getCount());
c.multiply(2);
System.out.println("After multiply(2): " + c.getCount());
c.divide(2);
System.out.println("After divide(2): " + c.getCount());
```

## HOMEWORK: Student Grade Tracker

```java
public class StudentGradeTracker {
    public static void main(String[] args) {
        System.out.println("=== Student Grade Tracker System ===\n");
        
        // Create first student
        System.out.println("Creating student: Emma Rodriguez");
        Student emma = new Student("Emma Rodriguez");
        System.out.println("Student created successfully!\n");
        
        // Add grades for Emma
        System.out.println("--- Adding Grades for Emma ---");
        emma.addGrade(95);
        System.out.println("Adding grade: 95 points");
        emma.addGrade(88);
        System.out.println("Adding grade: 88 points");
        emma.addGrade(92);
        System.out.println("Adding grade: 92 points");
        emma.addGrade(85);
        System.out.println("Adding grade: 85 points");
        
        // Print Emma's report
        System.out.println("\n--- Emma's Grade Report ---");
        emma.printReport();
        
        System.out.println("\n========================================\n");
        
        // Create second student
        System.out.println("Creating student: James Wilson");
        Student james = new Student("James Wilson");
        System.out.println("Student created successfully!\n");
        
        // Add grades for James
        System.out.println("--- Adding Grades for James ---");
        james.addGrade(78);
        System.out.println("Adding grade: 78 points");
        james.addGrade(82);
        System.out.println("Adding grade: 82 points");
        james.addGrade(75);
        System.out.println("Adding grade: 75 points");
        
        // Print James's report
        System.out.println("\n--- James's Grade Report ---");
        james.printReport();
        
        System.out.println("\n========================================\n");
        
        // Final summary
        System.out.println("Final Summary:");
        System.out.printf("Emma Rodriguez - Average: %.2f (%s)\n", emma.getAverage(), emma.getLetterGrade());
        System.out.printf("James Wilson - Average: %.2f (%s)\n", james.getAverage(), james.getLetterGrade());
    }
}

class Student {
    // Instance variables
    private String name;
    private int totalPoints;
    private int numAssignments;
    
    // Constructor
    public Student(String name) {
        this.name = name;
        this.totalPoints = 0;
        this.numAssignments = 0;
    }
    
    // Void method: adds a grade to the student's record
    public void addGrade(int points) {
        totalPoints += points;
        numAssignments++;
    }
    
    // Non-void method: calculates and returns average
    public double getAverage() {
        if (numAssignments == 0) {
            return 0.0;
        }
        return (double) totalPoints / numAssignments;
    }
    
    // Non-void method: returns letter grade based on average
    public String getLetterGrade() {
        double average = getAverage();
        if (average >= 90) {
            return "A";
        } else if (average >= 80) {
            return "B";
        } else if (average >= 70) {
            return "C";
        } else if (average >= 60) {
            return "D";
        } else {
            return "F";
        }
    }
    
    // Void method: prints a formatted grade report
    public void printReport() {
        System.out.println("Student Name: " + name);
        System.out.println("Total Points: " + totalPoints);
        System.out.println("Assignments Completed: " + numAssignments);
        System.out.printf("Current Average: %.2f\n", getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());
        
        // Add motivational status message
        String status;
        String letterGrade = getLetterGrade();
        if (letterGrade.equals("A")) {
            status = "Excellent work!";
        } else if (letterGrade.equals("B")) {
            status = "Good job!";
        } else if (letterGrade.equals("C")) {
            status = "Keep working hard!";
        } else if (letterGrade.equals("D")) {
            status = "Need improvement!";
        } else {
            status = "Seek help immediately!";
        }
        System.out.println("Status: " + status);
    }
}
```

```java
StudentGradeTracker.main(null);
```
