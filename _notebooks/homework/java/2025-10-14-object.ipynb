```java
// Book class with printInfo() method
class Book {
    String title;
    int pages;
    
    // Method to print book information
    void printInfo() {
        System.out.println("Title: " + title + ", Pages: " + pages);
    }
}

class MainPopcorn {
    public static void main(String[] args) {
        // Create one Book object
        Book myBook = new Book();
        
        // Set its fields
        myBook.title = "Java Programming";
        myBook.pages = 542;
        
        // Call printInfo() method
        myBook.printInfo();
    }
}

MainPopcorn.main(null);
```

```java
// Student class template
class Student {
    String name;
    int grade;
    int pets;
    int siblings;
    String favoriteSubject;  // Custom instance variable
    
    // Method to print student info
    void printInfo() {
        System.out.println("Name: " + name);
        System.out.println("Grade: " + grade);
        System.out.println("Pets: " + pets);
        System.out.println("Siblings: " + siblings);
        System.out.println("Favorite Subject: " + favoriteSubject);
        System.out.println();
    }
}

class Main {
    public static void main(String[] args) {
        // PART 1: Create three student objects
        
        Student student1 = new Student();
        student1.name = "Alex";
        student1.grade = 10;
        student1.pets = 2;
        student1.siblings = 1;
        student1.favoriteSubject = "Math";
        
        Student student2 = new Student();
        student2.name = "Jordan";
        student2.grade = 10;
        student2.pets = 0;
        student2.siblings = 3;
        student2.favoriteSubject = "English";
        
        Student student3 = new Student();
        student3.name = "Casey";
        student3.grade = 11;
        student3.pets = 5;
        student3.siblings = 2;
        student3.favoriteSubject = "Science";
        
        System.out.println("=== PART 1: Three Students ===");
        student1.printInfo();
        student2.printInfo();
        student3.printInfo();
        
        // PART 2: Create a reference variable with a different name
        // student3 goes by the nickname "C"
        Student C = student3;  // Both refer to the same object
        
        System.out.println("=== PART 2: Same Student, Different Name (Nickname) ===");
        System.out.println("Student by nickname 'C':");
        C.printInfo();
        
        // Proof they're the same object
        System.out.println("=== Proof they're the same object ===");
        System.out.println("Changing via 'C' nickname:");
        C.name = "Casey (Goes by C)";
        student3.printInfo();  // Changes are visible through student3 too!
    }
}

Main.main(null);
```
