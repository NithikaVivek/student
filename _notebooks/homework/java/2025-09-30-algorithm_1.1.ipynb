Hack 1: Activity chosen: Making a pizza from scratch

Algorithm:

1. Combine 3 cups of flour, 1 packet of yeast, 1 tablespoon of sugar, and 2 teaspoons of salt in a large mixing bowl
2. Add 1 cup of warm water and 2 tablespoons of olive oil to the dry ingredients and mix until a dough forms, then knead on a floured surface for 8-10 minutes until smooth
3. Place the dough in a greased bowl, cover with a damp towel, and let it rise in a warm spot for 1-2 hours until it doubles in size
4. While the dough rises, combine crushed tomatoes, minced garlic, olive oil, oregano, and salt in a saucepan and simmer for 15-20 minutes to make the sauce
5. Preheat the oven to 475°F and prepare toppings by shredding mozzarella cheese and slicing pepperoni or vegetables
6. Once the dough has risen, punch it down and stretch it into a 12-14 inch circle on a floured surface
7. Transfer the dough to a pizza pan, spread sauce on it leaving a 1-inch border, sprinkle cheese evenly, and add your toppings
8. Bake the pizza for 12-15 minutes until the crust is golden brown and the cheese is melted and bubbly
9. Remove from oven, let cool for 2-3 minutes, slice into 8 pieces, and serve immediately

```python
def calculate_grade(score1, score2, score3):
    """
    Calculate letter grade from three test scores
    
    Args:
        score1, score2, score3: Test scores (integers)
    
    Returns:
        grade: Letter grade (string)
    """
    # Step 2: Add the three scores together
    total = score1 + score2 + score3
    
    # Step 3: Divide by 3 to get average
    average = total / 3
    
    # Steps 4-8: Determine the letter grade using if-elif-else
    if average >= 90:
        grade = 'A'
    elif average >= 80:
        grade = 'B'
    elif average >= 70:
        grade = 'C'
    elif average >= 60:
        grade = 'D'
    else:
        grade = 'F'
    
    # Step 9: Display the grade
    return grade

# Test your function!
print("Test 1:", calculate_grade(95, 92, 88))  # Should be 'A'
print("Test 2:", calculate_grade(85, 80, 82))  # Should be 'B'
print("Test 3:", calculate_grade(75, 70, 72))  # Should be 'C'
print("Test 4:", calculate_grade(65, 60, 62))  # Should be 'D'
print("Test 5:", calculate_grade(55, 50, 52))  # Should be 'F'
```
