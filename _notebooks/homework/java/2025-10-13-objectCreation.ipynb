## Popcorn Hack #1: Stack vs Heap

1. Why does changing b not affect a, but changing array2 affects array1?
When you assign primitives (b = a), Java copies the actual value (10). So a and b are completely independent—changing b to 20 doesn't touch a. However, when you assign arrays (array2 = array1), Java copies only the reference (address) pointing to the array on the heap, not the actual data. Both array1 and array2 now point to the same array in memory, so modifying the array through either reference affects both.
2. Describe what's on the stack vs. the heap for this code.
Stack: Contains the variable names and values for primitives (a = 10, b = 20), plus the references (memory addresses) for array1 and array2. The stack stores fixed-size, simple data and addresses to objects.
Heap: Contains the actual array objects {1, 2, 3} and {99, 2, 3} (after modification). Only one array object exists in memory—both references point to the same location. When you modify array2[0] = 99, you're changing the data in that one array on the heap, which both references can access.

```java
public class MemoryDemo {
    public static void main(String[] args) {
        // Stack variables
        int a = 10;
        int b = a;  // Copy of value
        b = 20;     // Changing b doesn't affect a
        
        System.out.println("Primitives (Stack):");
        System.out.println("a = " + a);  // Still 10
        System.out.println("b = " + b);  // Now it's 20
        
        // Heap variables
        int[] array1 = {1, 2, 3};
        int[] array2 = array1;  // Copy of reference (address)
        array2[0] = 99;         // Changing array2 DOES affect array1
        
        System.out.println("\nArrays (Heap):");
        System.out.println("array1[0] = " + array1[0]);  // Now it's 99!
        System.out.println("array2[0] = " + array2[0]);  // Also 99
    }
}
MemoryDemo.main(null);
```

## Popcorn Hack #2
1. After haveBirthday(john) is called, what is John's age? Why?
John's age is 21. When you pass an object to a method, Java passes a copy of the reference (the address pointing to the object on the heap). Inside haveBirthday(), p points to the same Person object as john, so modifying p.age actually modifies the original object. The change persists after the method returns.
2. After reassignPerson(john) is called, what is John's name and age? Why?
John's name is still "John" and age is still 21. Even though reassignPerson() receives a copy of the reference, when you do p = new Person(...), you're only reassigning the local parameter p to point to a brand new object. This doesn't affect the original john reference outside the method—john still points to the original Person object. Reassigning a reference only changes where that local variable points, not the original.
3. Explain the difference between modifying an object's contents vs. reassigning a reference.
Modifying contents (p.age = p.age + 1): You're changing the data inside the object that the reference points to. Since both the local parameter and the original variable point to the same object, the change is visible outside the method.
Reassigning a reference (p = new Person(...)): You're making the local parameter point to a completely different object. This only affects the local parameter—the original reference outside the method is unchanged and still points to the original object.
```java
public class PersonDemo {
    static class Person {
        String name;
        int age;
        
        Person(String name, int age) {
            this.name = name;
            this.age = age;
        }
    }
    
    public static void haveBirthday(Person p) {
        p.age = p.age + 1;  // Modifying object content
        System.out.println("Inside method: " + p.name + " is now " + p.age);
    }
    
    public static void reassignPerson(Person p) {
        p = new Person("New Person", 99);  // Reassigning reference
        System.out.println("Inside reassign: " + p.name + " is " + p.age);
    }
    
    public static void main(String[] args) {
        Person john = new Person("John", 20);
        
        System.out.println("Before birthday: " + john.name + " is " + john.age);
        haveBirthday(john);
        System.out.println("After birthday: " + john.name + " is " + john.age);
        
        System.out.println("\nBefore reassign: " + john.name + " is " + john.age);
        reassignPerson(john);
        System.out.println("After reassign: " + john.name + " is " + john.age);
    }
}
PersonDemo.main(null);
```

## Homework Hack #1: Object Creation Practice

```java
public class ObjectCreation {
    public static void main(String[] args) {
        // 1. Create two Car objects using 'new'
        Car car1 = new Car("Tesla", 2024);
        Car car2 = new Car("Honda", 2022);
        
        // 2. Print each car's info
        System.out.println("=== Homework Hack #1: Object Creation ===");
        System.out.println(car1);
        System.out.println(car2);
        System.out.println();
    }
}

class Car {
    // 1. Declare variables: brand, year
    String brand;
    int year;
    
    // 2. Create a constructor to set those variables
    public Car(String brand, int year) {
        this.brand = brand;
        this.year = year;
    }
    
    // 3. Add a method or toString() to display car info
    @Override
    public String toString() {
        return "Car{" +
                "brand='" + brand + '\'' +
                ", year=" + year +
                '}';
    }
}
ObjectCreation.main(null);
```

## Homework Hack #2: Heap vs Stack Storage Demo

```java
public class HeapVsStack {
    public static void main(String[] args) {
        System.out.println("=== Homework Hack #2: Heap vs Stack Demo ===\n");
        
        // 1. Create a primitive variable (int pages)
        int pages = 300;
        System.out.println("PRIMITIVES (Stored on Stack):");
        System.out.println("Original: pages = " + pages);
        
        // 2. Create another primitive variable that copies it
        int pagesCopy = pages;
        System.out.println("Copy: pagesCopy = " + pagesCopy);
        
        // 3. Change the original primitive
        pages = 500;
        System.out.println("\nAfter changing pages to 500:");
        System.out.println("pages = " + pages);
        System.out.println("pagesCopy = " + pagesCopy);
        System.out.println("→ pagesCopy stayed 300! They are independent copies.\n");
        
        // 4. Create a Book object (Book b1)
        System.out.println("OBJECTS (Reference stored on Stack, Object on Heap):");
        Book b1 = new Book("Java Basics");
        System.out.println("Original: b1 = " + b1);
        
        // 5. Create another Book reference that copies it
        Book b2 = b1;
        System.out.println("Copy: b2 = " + b2);
        
        // 6. Change the original Book title
        b1.setTitle("Advanced Java");
        System.out.println("\nAfter changing b1's title to 'Advanced Java':");
        System.out.println("b1 = " + b1);
        System.out.println("b2 = " + b2);
        System.out.println("→ b2 changed too! They both point to the SAME object on the heap.");
    }
}

class Book {
    // 1. Declare variable: String title
    private String title;
    
    // 2. Create a constructor to set the title
    public Book(String title) {
        this.title = title;
    }
    
    // Helper method to change the title
    public void setTitle(String newTitle) {
        this.title = newTitle;
    }
    
    // 3. Create a toString() to show the title
    @Override
    public String toString() {
        return "Book{" +
                "title='" + title + '\'' +
                '}';
    }
}
HeapVsStack.main(null);
```
