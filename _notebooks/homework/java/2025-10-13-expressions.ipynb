## Popcorn Hack 2
**Formatting tools used in Menu.java:**

1. **`System.out.println()`** - Used to print each menu line and automatically move to the next line, creating a clean vertical menu layout
2. **`System.out.print()`** - Used for "Choose an option: " to keep the cursor on the same line, allowing user input to appear right after the prompt
3. **String concatenation with `+`** - Combines strings with variables (like `"You selected option: " + choice`) to create dynamic output
4. **Equals signs (`====`)** - Creates a visual header/border for the menu title

**How they make output cleaner/better:**
- The consistent use of `println()` creates organized, easy-to-read vertical spacing
- `print()` for the prompt makes the input feel more interactive and natural
- The header with equal signs visually separates the menu from other output
- String concatenation allows dynamic content while maintaining readable formatting

### Menu Hack:

````java
import java.util.Scanner;

public class Menu {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Main Menu ====");
        System.out.println("1. Start Game");
        System.out.println("2. Instructions");
        System.out.println("3. Exit");
        System.out.println("4. Settings");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.println("You selected option: " + choice);

        int optionCount = 4;
        System.out.println("There are " + optionCount + " total options.");
    }
}

Menu.main(null);
````

## Homework Hack 2

````java
import java.util.Scanner;

public class CalculatorMenu {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("==== Calculator Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.print("Enter first number: ");
        double num1 = sc.nextDouble();

        System.out.print("Enter second number: ");
        double num2 = sc.nextDouble();

        double result = 0;
        String operation = "";

        if (choice == 1) {
            result = num1 + num2;
            operation = "Addition";
        } else if (choice == 2) {
            result = num1 - num2;
            operation = "Subtraction";
        } else if (choice == 3) {
            result = num1 * num2;
            operation = "Multiplication";
        } else if (choice == 4) {
            if (num2 != 0) {
                result = num1 / num2;
                operation = "Division";
            } else {
                System.out.println("Error: Cannot divide by zero!");
                return;
            }
        } else {
            System.out.println("Invalid option!");
            return;
        }

        System.out.println("\n" + operation + " Result: " + result);
    }
}

CalculatorMenu.main(null);
````
