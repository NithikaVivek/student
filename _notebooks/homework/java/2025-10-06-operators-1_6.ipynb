## Popcorn Hack #1: Transform Beginner Code

Below, the beginner code is rewritten using compound assignment operators for clarity and efficiency.

```java
// Popcorn Hack #1: Compound Assignment Operators
int playerScore = 1000;
int playerHealth = 100;
int enemiesDefeated = 0;

// Player defeats an enemy worth 250 points
playerScore += 250; // Add 250 to score
System.out.println("Score after enemy: " + playerScore);

// Player takes 15 damage
playerHealth -= 15; // Subtract 15 from health
System.out.println("Health after damage: " + playerHealth);

// Enemy count goes up
enemiesDefeated++; // Increment enemy count
System.out.println("Enemies defeated: " + enemiesDefeated);

// Boss battle: double the current score!
playerScore *= 2; // Double the score
System.out.println("Score after boss: " + playerScore);

// Healing potion restores health to 80% of current
playerHealth *= 4;
playerHealth /= 5; // Integer math for 80%
System.out.println("Health after potion: " + playerHealth);
```

## Popcorn Hack #2: Score Updates with Compound Assignment

Write a short program where a variable called score starts at 100. Use at least three different compound assignment operators to update score, and print the result after each step.

```java
// Popcorn Hack #2: Score Updates
int score = 100;
System.out.println("Starting score: " + score);

score -= 20; // Deduct points for a wrong answer
System.out.println("After wrong answer (-20): " + score);

score *= 2; // Double the score with a power-up
System.out.println("After power-up (*2): " + score);

score %= 7; // Find remainder after dividing by 7
System.out.println("After modulo 7 (%=): " + score);
```

# Homework Assignment: Social Media Influencer Simulator (Option A)

Below is a simulation using at least three different compound assignment operators, meaningful variable names, print statements, and comments explaining each operation.

## Option A: Social Media Influencer Simulator

Model the journey of a content creator tracking their growth:


```java
// Social Media Influencer Simulator
System.out.println("=== SOCIAL MEDIA SIMULATOR ===");

int followers = 500; // Starting followers
int posts = 10;      // Number of posts made
int engagement = 35; // Likes/comments received
int position = 400;  // Ranking position
int sponsorshipEarnings = 0; // Money earned from sponsors

System.out.println("Starting followers: " + followers);
System.out.println("Posted a new video!");
followers += 250; // Gain followers from viral video
System.out.println("Followers: " + followers + " (+250 from the viral video)");

System.out.println("Controversial opinion posted...");
followers -= 50; // Lose followers
System.out.println("Followers: " + followers + " (-50 from upset followers)");

System.out.println("Trending hashtag boost!");
followers *= 2; // Double followers from trending
System.out.println("Followers: " + followers + " (doubled from trending!)");

engagement *= 2; // Double engagement
System.out.println("Engagement after trending: " + engagement);

engagement /= posts; // Average engagement per post
System.out.println("Average engagement per post: " + engagement);

position %= 100; // Find ranking position
System.out.println("Final ranking position: " + position);

System.out.println("=== SIMULATION COMPLETE ===");
```
